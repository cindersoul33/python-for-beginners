```python
#comparison operators
a = 2
b = 3

print(a == b) #False
print(a != b) #True
print(a > b)
print(a <= b)
```

```python
#boolean operators
#toán tử boolean dùng để kết hợp hoặc phủ định các giá trị True/False
#trả về kết quả True hoặc Fales
#3 toán tử boolean chính: and, or, not

condition1 = True
condition2 = False

print(not condition1)
print(condition1 and condition2)
print(condition1 or condition2)
```
