```python
def is_aldult(age):
    if age > 18:
        return True
    else:
        return False

print(is_aldult(15))
```

```python
def is_aldult(age):
    return True if age > 18 else False

print(is_aldult(15))
```

```python
def is_aldult(age):
    return age > 18

print(is_aldult(15))
```
